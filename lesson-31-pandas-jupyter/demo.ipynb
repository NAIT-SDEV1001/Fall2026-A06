```python
import pandas as pd
import matplotlib.pyplot as plt
```

# Reading Data in Jupyter Notebook

1) Reading a CSV

```python
progression_data = pd.read_csv("rpg_level_progression.csv")

print(f"(Rows, Column): {progression_data.shape}")
progression_data.head()
```

```python
progression_data.info()
print(progression_data.describe())
```

## Finding Unique Stat Gains

Looking for how many unique stat gains in defense

```python
unique_defense_gains = progression_data["defense_gain"].nunique()
unique_defense_gains
```

## Sorting By HP Gain

```python
top_hp_gains = progression_data.sort_values(by="hp_gain", ascending=False)
print(top_hp_gains.iloc[0])
top_hp_gains.head()
```

```python
progression_data.value_counts()
```

```python
progression_data.isna().sum()
```

```python
progression_data.columns = progression_data.columns.str.replace(' ', '')

plt.plot(progression_data["level"], progression_data["xp_to_next_level"])
plt.xlabel("Level")
plt.ylabel("XP to Next Level")
plt.show()
```

```python
progression_data[['xp_to_next_level', 'hp_gain', 'attack_gain', 'defense_gain']].plot(
    kind='bar',
    stacked=True
)

plt.xlabel("Level")
plt.ylabel('Value')
```
