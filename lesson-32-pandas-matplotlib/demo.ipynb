```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("jupyter_pandas_matplotlib_example_start/data/Property_Assessment_Data_(Current_Calendar_Year)_20251030.csv")

data.head()
```

```python
residential_data = data[data["Tax Class"] == "Residential"]

residential_data.head()
```

```python
top_5_most_expensive = residential_data.sort_values(by="Assessed Value", ascending=False)[:5]

top_5_most_expensive.head(10)
```

```python
index_of_max = residential_data["Assessed Value"].idxmax()
print(index_of_max)
most_expensive = residential_data.loc[index_of_max]
most_expensive
```

```python
top_5_most_expensive["Property Address"] = top_5_most_expensive["House Number"].astype(str) + ' ' + top_5_most_expensive['Street Name']

top_5_most_expensive.head()
```

```python
property_names = top_5_most_expensive["Property Address"]
assessed_values = top_5_most_expensive["Assessed Value"]

plt.figure(figsize=(10, 6))
plt.bar(property_names, assessed_values)
plt.xlabel('Property Address')
plt.ylabel('Assessed Value')
plt.title('Top 5 Most Expensive Residential Properties')
plt.xticks(rotation=45, ha='right')  # Rotate x labels for better readability
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.plot(property_names, assessed_values, marker='o', linestyle='-', color='orange')
plt.xlabel('Property Address')
plt.ylabel('Assessed Value')
plt.title('Top 5 Most Expensive Residential Properties - Line Chart')
plt.xticks(rotation=45, ha='right')  # Rotate x labels for better readability
plt.grid(False)
plt.show()
```

```python
downtown = data[data['Neighbourhood'] == 'DOWNTOWN']

plt.figure(figsize=(10, 6))
plt.scatter(downtown['Latitude'], downtown['Assessed Value'], alpha=0.5, color='purple')
plt.xlabel('Latitude')
plt.ylabel('Assessed Value')
plt.title('Downtown Properties: Assessed Value vs Latitude')
plt.show()
```
